```python
import pandas as pd
data= pd.read_csv('C:\\Users\\knowl\\.cache\\kagglehub\\datasets\\abdallahwagih\\spam-emails\\versions\\1\\spam.csv')
data[:3]
```

```python
target=data['Category']
feature=data['Message']
```

```python

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(feature, target, test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
X1 = vectorizer.fit_transform(X_train)
X2 = vectorizer.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

clf = RandomForestClassifier(max_depth=50, random_state=0,n_estimators=300)

clf.fit(X1,y_train)

pred=clf.predict(X2)

accuracy_score(y_test, pred)
```

```python
from joblib import Parallel, delayed 
import joblib 


# Save the model as a pickle in a file 
joblib.dump(clf, 'rf.pkl') 



```

```python
# Load the model from the file 
rfmodel = joblib.load('rf.pkl') 
  
# Use the loaded model to make predictions 
rfmodel.predict(X2) 
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

neigh = KNeighborsClassifier(n_neighbors=3)

neigh.fit(X1,y_train)

pred=neigh.predict(X2)

accuracy_score(y_test, pred)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

clf = LogisticRegression(random_state=0)

clf.fit(X1,y_train)

pred=clf.predict(X2)

accuracy_score(y_test, pred)
```

```python
from sklearn.svm import SVC
clf= SVC(gamma='auto')

clf.fit(X1,y_train)

pred=clf.predict(X2)

accuracy_score(y_test, pred)
```

```python
from sklearn.tree import DecisionTreeClassifier
clf = DecisionTreeClassifier(random_state=0)

clf.fit(X1,y_train)

pred=clf.predict(X2)

accuracy_score(y_test, pred)
```
