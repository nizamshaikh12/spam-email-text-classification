```python
import pandas as pd
data= pd.read_csv('/kaggle/input/spam-emails/spam.csv')
data[:3]
```

```python
len(data)
```

```python
target=data['Category']
feature=data['Message']
```

```python

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(feature, target, test_size=0.20, random_state=42)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
vectorizer = TfidfVectorizer()
X1 = vectorizer.fit_transform(X_train)
X2 = vectorizer.transform(X_test)
```

```python
from sklearn.ensemble import RandomForestClassifier
clf = RandomForestClassifier(max_depth=50, random_state=0,n_estimators=300)

```

```python
clf.fit(X1,y_train)
```

```python
pred=clf.predict(X2)
```

```python
from collections import Counter
Counter(pred)
```
